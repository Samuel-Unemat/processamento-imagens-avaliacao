```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
repo_url = "https://github.com/Samuel-Unemat/processamento-imagens-avaliacao"

!git clone {repo_url}
```

```python
base_path = "processamento-imagens-avaliacao/imagem"

# Lista todas as imagens da pasta
imagens = os.listdir(base_path)

print("Imagens encontradas:", imagens)
```

```python
img_path = os.path.join(base_path, "imagem1.jpg")

img = cv2.imread(img_path)

# Verificação
if img is None:
    print("Erro ao carregar a imagem. Verifique o nome/caminho.")
else:
    print("Imagem carregada com sucesso!")
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.title("Imagem carregada do GitHub")
plt.axis('off')
```
